```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

#1. Traemos a la Mona Lisa desde la Wikipedia
!wget "https://upload.wikimedia.org/wikipedia/commons/thumb/e/ec/Mona_Lisa%2C_by_Leonardo_da_Vinci%2C_from_C2RMF_retouched.jpg/402px-Mona_Lisa%2C_by_Leonardo_da_Vinci%2C_from_C2RMF_retouched.jpg" -O monalisa.jpg

#2. Leemos la imagen como una MATRIZ de numeros
img = mpimg.imread('monalisa.jpg')

#3. Datos de la autopsia digital
print(f"Tipo de Dato: {type(img)}")
print(f"shape (Dimensiones): {img.shape}")

#4. Mostramos al paciente
plt.imshow(img)
plt.axis('off')
plt.show()
```

```python
#1. Definicion del Vector

Pesos = np.array([0.299, 0.587, 0.114])

#2. Se multiplica la Matriz (img) por el vector (Pesos)
mona_lisa_bn = img @ Pesos

#3. Muestra el resultado
plt.imshow(mona_lisa_bn, cmap='gray')
plt.axis('off')
plt.show()
```

```python
# recorte = matriz[ fila_inicio : fila_fin , columna_inicio : columna_fin ]

cara = mona_lisa_bn[50:250, 100:300]
plt.imshow(cara, cmap='gray')
plt.axis('off')
plt.show()
```

```python
#1. Buscar el valor mas brillante que tiene la cara actualmente
max_valor = cara.max()
print(f"El valor maximo es: {max_valor}")

#2. Operacion matematica: Restamos a ese maximo tu matriz 'cara'
#(Numpy es tan listo que restara pixel por pixel automaticamente)

evil_mona = max_valor - cara

#3. A revelar al monstruo
plt.imshow(evil_mona, cmap='gray')
plt.axis('off')
plt.show()
```
